```python
import random

# Generate a random number between 1 and 10[cite: 1]
secret_number = random.randint(1, 10)

# Set the maximum number of attempts allowed to 3[cite: 1]
attempts = 3

print("Setup is ready! A secret number between 1 and 10 has been chosen.")
```

```python
# Start the guessing game loop
print("--- Let's Play the Number Guessing Game ---")

while attempts > 0:
    try:
        guess = int(input("Guess the number (between 1 and 10): "))
    except ValueError:
        print("Invalid input! Please enter a valid number.")
        continue

    # Check if the guess is out of the valid range
    if guess < 1 or guess > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue  # Skips the rest and goes back to the loop

    # Check if the guess is too high or too low
    if guess > secret_number:
        print("Too high. Try again.")
    elif guess < secret_number:
        print("Too low. Try again.")
    else:
        print("Congratulations! You guessed the correct number.")
        break  # Exits the loop when correct

    attempts -= 1
    print(f"Attempts remaining: {attempts}")

else:
    print(f"Better luck next time! The secret number was {secret_number}.")
```

```python
print("--- Multiplication Table Generator ---")

# Prompt user for input
number = int(input("Enter the number for which you want the multiplication table: "))
print(f"Great! Let's generate the table for {number}.")
```

```python
# Generate the Multiplication Table using a for loop from 1 to 10
for i in range(1, 11):
    result = number * i
    # Display each line in the format: "number x i = result"[cite: 3]
    print(f"{number} x {i} = {result}")
```

```python
print("--- Body Mass Index (BMI) Calculator ---")

# Define a function calculate_bmi(weight, height) that returns the BMI[cite: 3]
def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    return bmi

print("BMI calculation tool is ready!")
```

```python
# Prompt the user for their weight (in kg) and height (in meters)[cite: 3]
weight_input = float(input("Enter your weight in kg: "))
height_input = float(input("Enter your height in meters: "))

# Use the function to calculate and display the BMI[cite: 3]
final_bmi = calculate_bmi(weight_input, height_input)

# Display the result rounded to two decimal places
print(f"Your BMI is: {final_bmi:.2f}")
```
